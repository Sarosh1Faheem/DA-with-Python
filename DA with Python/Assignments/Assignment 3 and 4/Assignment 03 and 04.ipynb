## List

1. Write a program that accepts a list from user and print the alternate element of list.

````python
numbers = input("Enter list elements separated by space: ").split()

print("Alternate elements are:")

for i in range(0, len(numbers), 2):
    print(numbers[i])
````

2. Write a program that accepts a list from user. Your program should reverse the content of list and
display it. Do not use reverse() method.

````python
numbers = input("Enter list elements separated by space: ").split()

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed list:", reversed_list)
````

3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard.

````python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number is:", largest)
````

4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

````python
numbers = input("Enter list elements separated by space: ").split()

last_element = numbers[-1]

for i in range(len(numbers) - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last_element

print("Rotated list:", numbers)
````

5. Write a program that input a string and ask user to delete a given word from a string.

````python
sentence = input("Enter a sentence: ")
word = input("Enter the word you want to delete: ")

words = sentence.split()

new_words = []

for item in words:
    if item != word:
        new_words.append(item)

result = " ".join(new_words)

print("Updated sentence:", result)
````

7. Write a program with a function that accepts a string from keyboard and create a new string after
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses"

````python
def capitalize_words(sentence):
    words = sentence.split()
    new_sentence = []

    for word in words:
        new_sentence.append(word.capitalize())

    return " ".join(new_sentence)


sentence = input("Enter a sentence: ")

result = capitalize_words(sentence)

print(result)
````

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :
``` 
2 11 7 12
5 2  9 15
8 8 10 42
```
```
Sum of Row 1 = 32
Sum of Row 2 = 31
Sum of Row 3 = 63
```

````python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix = []

for i in range(rows):
    row = list(map(int, input(f"Enter elements of row {i + 1}: ").split()))
    matrix.append(row)


for i in range(rows):
    row_sum = 0

    for j in range(columns):
        row_sum = row_sum + matrix[i][j]

    print(f"Sum of row {i + 1} =", row_sum)
````

9. Write a program to add two matrices of size n x m.

````python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []


# Input Matrix 1
print("Enter elements of Matrix 1:")

for i in range(rows):
    row = list(map(int, input().split()))
    matrix1.append(row)


# Input Matrix 2
print("Enter elements of Matrix 2:")

for i in range(rows):
    row = list(map(int, input().split()))
    matrix2.append(row)


# Add Matrices
result = []

for i in range(rows):

    row = []

    for j in range(columns):

        addition = matrix1[i][j] + matrix2[i][j]

        row.append(addition)

    result.append(row)


# Display Matrix 1
print("\nMatrix 1:")

for row in matrix1:
    print(row)


# Display Matrix 2
print("\nMatrix 2:")

for row in matrix2:
    print(row)


# Display Result
print("\nResultant Matrix:")

for row in result:
    print(row)
````

10. Write a program to multiply two matrices

````python
rows1 = int(input("Enter number of rows of Matrix 1: "))
columns1 = int(input("Enter number of columns of Matrix 1: "))

matrix1 = []

print("Enter elements of Matrix 1:")

for i in range(rows1):
    row = list(map(int, input().split()))
    matrix1.append(row)


rows2 = int(input("Enter number of rows of Matrix 2: "))
columns2 = int(input("Enter number of columns of Matrix 2: "))

matrix2 = []

print("Enter elements of Matrix 2:")

for i in range(rows2):
    row = list(map(int, input().split()))
    matrix2.append(row)


# Display Matrix 1
print("\nMatrix 1:")

for row in matrix1:
    print(row)


# Display Matrix 2
print("\nMatrix 2:")

for row in matrix2:
    print(row)


# Check whether multiplication is possible
if columns1 != rows2:
    print("\nMatrix multiplication is not possible.")

else:
    
    result = []

    for i in range(rows1):
        
        row = []

        for j in range(columns2):
            
            total = 0

            for k in range(columns1):
                total = total + matrix1[i][k] * matrix2[k][j]

            row.append(total)

        result.append(row)


    # Display Resultant Matrix
    print("\nResultant Matrix:")

    for row in result:
        print(row)
````

##### 4 - Count Frequency of Numbers

`numbers = [1,2,2,3,3,3,4,4,4,4]`

Create frequency dictionary.

````python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] = frequency[number] + 1
    else:
        frequency[number] = 1

print(frequency)
````

##### 5 - Word Counter

Take sentence input and count word frequency

````python
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print(frequency)
````

##### 6 - Student Marks & Average

```
marks = {"Math":80,
         "English":70,
         "Physics":90
         }
```

Calculate total and average.

````python
marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = 0

for mark in marks.values():
    total = total + mark

average = total / len(marks)

print("Total Marks:", total)
print("Average Marks:", average)
````

##### 7 - Highest Scoring Student
```
students = {
          "Ali":85,
          "Sara":92,
          "Ahmed":78,
          "Zara":95
          }
```
Find highest scoring student.

````python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = ""
highest_marks = 0

for name, marks in students.items():

    if marks > highest_marks:
        highest_marks = marks
        highest_student = name


print("Highest Scoring Student:", highest_student)
print("Marks:", highest_marks)
````
