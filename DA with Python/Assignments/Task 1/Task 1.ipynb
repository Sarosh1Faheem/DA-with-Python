### Scenario

A school wants a small Python program to manage its people. Everyone in the school is a Person, but students, teachers, and admin staff each have their own details and behavior.
_Requirements_
#### Part 1: Base class (10 marks)
* Create a class Person with:
* Attributes: name, age, cnic
* Method display_info() that prints name, age, and CNIC
* Method role() that returns "Person"

#### Part 2: Child classes (20 marks)
* Create three classes that inherit from Person. Use super() in every constructor.
##### Student
    * Extra attributes: roll_no, marks (a list of numbers)
* Method average() that returns the average of marks
* Method grade() that returns A (80+), B (70 to 79), C (60 to 69), or F (below 60)
* Override role() to return "Student"
##### Teacher
    * Extra attributes: subject, salary
    * Method annual_salary() that returns salary × 12
    * Override role() to return "Teacher"
##### Admin
* Extra attributes: department, salary
* Method annual_salary() that returns salary × 12 plus a 10% bonus
* Override role() to return "Admin"
* Each child class must override display_info(), call super().display_info() first, then print its own extra details.

#### Part 3: Polymorphism (10 marks)
* Create a list with at least 2 students, 2 teachers, and 1 admin
* Loop through the list and call role() and display_info() on each object
* Print a line of dashes between each person

#### Part 4: Reports (10 marks)
* Using isinstance(), write code that:
* Counts how many students, teachers, and admins there are
* Prints the name of the student with the highest average
* Prints the total annual salary paid to all staff (teachers and admins)

```python
# ==========================================
# Part 1: Base Class - Person
# ==========================================

class Person:
    def __init__(self, name, age, cnic):
        self.name = name
        self.age = age
        self.cnic = cnic

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}, CNIC: {self.cnic}")

    def role(self):
        return "Person"


# ==========================================
# Part 2: Child Classes
# ==========================================

# Student Class
class Student(Person):
    def __init__(self, name, age, cnic, roll_no, marks):
        super().__init__(name, age, cnic)
        self.roll_no = roll_no
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()

        if avg >= 80:
            return "A"
        elif avg >= 70:
            return "B"
        elif avg >= 60:
            return "C"
        else:
            return "F"

    def role(self):
        return "Student"

    def display_info(self):
        super().display_info()
        print(f"Roll No: {self.roll_no}")
        print(f"Average: {self.average():.1f}, Grade: {self.grade()}")


# Teacher Class
class Teacher(Person):
    def __init__(self, name, age, cnic, subject, salary):
        super().__init__(name, age, cnic)
        self.subject = subject
        self.salary = salary

    def annual_salary(self):
        return self.salary * 12

    def role(self):
        return "Teacher"

    def display_info(self):
        super().display_info()
        print(f"Subject: {self.subject}")
        print(f"Annual Salary: {self.annual_salary()}")


# Admin Class
class Admin(Person):
    def __init__(self, name, age, cnic, department, salary):
        super().__init__(name, age, cnic)
        self.department = department
        self.salary = salary

    def annual_salary(self):
        # Monthly salary * 12 + 10% bonus
        return (self.salary * 12) * 1.10

    def role(self):
        return "Admin"

    def display_info(self):
        super().display_info()
        print(f"Department: {self.department}")
        print(f"Annual Salary: {self.annual_salary()}")


# ==========================================
# Part 3: Polymorphism
# ==========================================

people = [
    Student(
        "Ali",
        18,
        "42101-1234567-1",
        101,
        [85, 85, 85]
    ),

    Student(
        "Sara",
        19,
        "42101-2222222-4",
        102,
        [75, 78, 80]
    ),

    Teacher(
        "Sir Ahmed",
        35,
        "42101-7654321-3",
        "Python",
        100000
    ),

    Teacher(
        "Miss Hina",
        32,
        "42101-8888888-5",
        "Mathematics",
        90000
    ),

    Admin(
        "Sana",
        29,
        "42101-1111111-2",
        "Accounts",
        60000
    )
]


# Display information using polymorphism
for person in people:
    print(f"Role: {person.role()}")
    person.display_info()
    print("-" * 30)


# ==========================================
# Part 4: Reports using isinstance()
# ==========================================

student_count = 0
teacher_count = 0
admin_count = 0

top_student = None
total_staff_salary = 0


for person in people:

    # Count Students
    if isinstance(person, Student):
        student_count += 1

        # Find student with highest average
        if top_student is None or person.average() > top_student.average():
            top_student = person

    # Count Teachers
    elif isinstance(person, Teacher):
        teacher_count += 1
        total_staff_salary += person.annual_salary()

    # Count Admins
    elif isinstance(person, Admin):
        admin_count += 1
        total_staff_salary += person.annual_salary()


# Print Reports
print(f"Students: {student_count}, Teachers: {teacher_count}, Admins: {admin_count}")

print(
    f"Top Student: {top_student.name} "
    f"({top_student.average():.1f})"
)

print(f"Total Staff Salary: {total_staff_salary}")
```
