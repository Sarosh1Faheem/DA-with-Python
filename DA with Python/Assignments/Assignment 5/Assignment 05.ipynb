### Q1 Working with List

1. Create a list called numbers containing at least 8 integers.
2.	Using a for loop, do the following:
    * Print all numbers in the list.
    * Print the sum of all numbers.
    * Print only the even numbers.

_Requirements:_
* Use at least one loop.
* Use conditional statements.
* Display clear output.

````python
# Create a list containing at least 8 integers
numbers = [10, 25, 14, 33, 8, 41, 20, 17]

print("Numbers in the list:")

# Print all numbers using a for loop
for number in numbers:
    print(number)

# Calculate and print the sum of all numbers
total = 0

for number in numbers:
    total = total + number

print("\nSum of all numbers:", total)

# Print only the even numbers
print("\nEven numbers:")

for number in numbers:
    if number % 2 == 0:
        print(number)
````

### Q2 Dictionary Basics

1.	Create a dictionary called student with the following keys:
    * name
    * age
    * course
    * marks (must store a list of 3 marks)
2.	Using loops:
    * Print all keys and their values.
    * Calculate the average of the marks.
    * Display whether the student Passed (average ≥ 50) or Failed.

_Requirements:_
* Dictionary must contain a list.
* Use loops to calculate the average.

````python
# Create a dictionary containing student information
student = {
    "name": "Ali",
    "age": 20,
    "course": "Python Programming",
    "marks": [75, 68, 82]
}

# Print all keys and their values using a loop
print("Student Details:")

for key, value in student.items():
    print(key, ":", value)

# Calculate the average of marks using a loop
total_marks = 0

for mark in student["marks"]:
    total_marks = total_marks + mark

average = total_marks / len(student["marks"])

print("\nAverage Marks:", average)

# Check whether the student passed or failed
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
````

### Q3 List of Dictionaries

1.	Create a list called employees.
2.	Each employee must be stored as a dictionary containing:
    * name
    * department
    * salary
3.	Using loops:
    * Print details of all employees.
    * Find the employee with the highest salary.
    * Calculate the total salary expense.

_Requirements:_
* Minimum 3 employees.
* Use comparison logic.
* Use loops to process data.


````python
# Create a list of dictionaries containing employee information
employees = [
    {
        "name": "Ahmed",
        "department": "IT",
        "salary": 60000
    },
    {
        "name": "Sara",
        "department": "HR",
        "salary": 55000
    },
    {
        "name": "Usman",
        "department": "Finance",
        "salary": 75000
    }
]

# Print details of all employees
print("Employee Details:")

for employee in employees:
    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print("----------------------")

# Find the employee with the highest salary
highest_salary = employees[0]

for employee in employees:
    if employee["salary"] > highest_salary["salary"]:
        highest_salary = employee

print("\nEmployee with Highest Salary:")
print("Name:", highest_salary["name"])
print("Department:", highest_salary["department"])
print("Salary:", highest_salary["salary"])

# Calculate the total salary expense
total_salary = 0

for employee in employees:
    total_salary = total_salary + employee["salary"]

print("\nTotal Salary Expense:", total_salary)
````

### Q4 While Loops with user Input

1.	Ask the user to enter numbers.
2.	Store the numbers in a list.
3.	Stop when the user enters -1.
4.	After the loop ends:
    * Print the complete list.
    * Print the largest number.
    * Print the smallest number.

_Requirements:_
* Must use a while loop.
* Must store input in a list.
* Handle user input correctly.


````python
# Create an empty list to store numbers
numbers = []

print("Enter numbers one by one.")
print("Enter -1 to stop.")

# Use a while loop to take input
while True:
    number = int(input("Enter a number: "))

    # Stop the loop when user enters -1
    if number == -1:
        break

    # Store the number in the list
    numbers.append(number)

# Display the complete list
print("\nComplete List:", numbers)

# Check whether the user entered any numbers
if len(numbers) > 0:
    # Print the largest number
    print("Largest Number:", max(numbers))

    # Print the smallest number
    print("Smallest Number:", min(numbers))
else:
    print("No numbers were entered.")
````

### Q5 Word Frequency Number

1.	Ask the user to enter a sentence.
2.	Convert the sentence into a list of words.
3.	Use a dictionary to count how many times each word appears.
4.	Print the word frequency clearly.
Example Format (for understanding only):
apple: 3
banana: 2
orange: 1

_Requirements:_
* Use string splitting.
* Use a dictionary for counting.
* Use loops for processing.


````python
# Ask the user to enter a sentence
sentence = input("Enter a sentence: ")

# Convert the sentence into a list of words
words = sentence.lower().split()

# Create an empty dictionary for word frequency
word_frequency = {}

# Count each word using a loop
for word in words:
    if word in word_frequency:
        word_frequency[word] = word_frequency[word] + 1
    else:
        word_frequency[word] = 1

# Display the word frequency
print("\nWord Frequency:")

for word, frequency in word_frequency.items():
    print(word, ":", frequency)
````

Using nested loops, print a multiplication table from 1 to 5.


_Requirements:_

````python
# QUESTION 6: Nested Loops - Multiplication Table

# Print multiplication table from 1 to 5
print("Multiplication Table:")

# Outer loop
for i in range(1, 6):

    # Inner loop
    for j in range(1, 6):
        print(i * j, end="\t")

    # Move to the next line after each row
    print()
````
