1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd

# Employee data used in Questions 1 and 4-9
data = {
    "Name": ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Sana"],
    "Age": [28, 35, 42, 31, 26, 38],
    "City": ["Delhi", "Mumbai", "Delhi", "Bengaluru", "Delhi", "Mumbai"],
    "Salary": [42000, 55000, 68000, 61000, 39000, 72000],
}
df = pd.DataFrame(data)

print("First 3 rows:")
print(df.head(3))
print("\nShape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

2) Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
df_updated = df.copy()
df_updated["Bonus"] = df_updated["Salary"] * 0.10
df_updated["Total"] = df_updated["Salary"] + df_updated["Bonus"]
df_updated = df_updated.rename(columns={"City": "Location"})
df_updated = df_updated.drop(columns=["Bonus"])

print(df_updated)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
missing_df = pd.DataFrame({
    "Name": ["Aarav", "Diya", None, "Meera", "Rohan"],
    "Age": [28, None, 42, 31, 26],
    "Salary": [42000, 55000, None, 61000, 39000],
    "City": ["Delhi", "Mumbai", "Delhi", None, "Delhi"],
})

print("Missing values per column:")
print(missing_df.isna().sum())

missing_df[["Age", "Salary"]] = missing_df[["Age", "Salary"]].fillna(
    missing_df[["Age", "Salary"]].mean()
)
clean_df = missing_df.dropna()
print("\nAfter filling numeric values with their column means and dropping remaining missing rows:")
print(clean_df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_df = df.sort_values(
    by=["Salary", "Age"],
    ascending=[False, True],
)
print("Sorted by Salary descending, then Age ascending:")
print(sorted_df)

print("\nTop 3 highest-paid employees:")
print(df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
salary_by_city = df.groupby("City")["Salary"].agg(
    Average_Salary="mean",
    Employee_Count="count",
    Maximum_Salary="max",
)
print(salary_by_city)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame({
    "Name": ["Aarav", "Diya", "Kabir", "Meera", "Rohan"],
    "DeptID": [1, 2, 1, 3, 4],
})
departments = pd.DataFrame({
    "DeptID": [1, 2, 3],
    "Department": ["Engineering", "Sales", "HR"],
})

print("Inner join:")
print(employees.merge(departments, on="DeptID", how="inner"))
print("\nLeft join:")
print(employees.merge(departments, on="DeptID", how="left"))

employees_extra = pd.DataFrame({
    "Name": ["Sana", "Aarav"],
    "DeptID": [2, 1],
})
combined = pd.concat([employees, employees_extra], ignore_index=True).drop_duplicates()
print("\nConcatenated employees with duplicate rows removed:")
print(combined)
```
