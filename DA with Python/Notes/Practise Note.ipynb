3. Completes the following steps of small task:

```python
eng = 87
isl = 85
math = 90
total_marks = 300
obt_marks = eng + isl + math
print("Marks Obtained in English:", eng)
print("Marks Obtained in Islamiat:", isl)
print("Marks Obtained in Mathematics:", math)
print("Total Marks Obtained:", obt_marks)
print("Total Marks:", total_marks)
print("Percentage:", (obt_marks / total_marks) * 100, "%")
```

4. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
pi = 3.14159
radius = float(input("Enter the radius of the circle: "))

area = pi * (radius ** 2)
circumference = 2 * pi * radius

print("Your Radius is", radius)
print("Area of Circle:", area)
print("Circumference of Circle:", circumference)
```
